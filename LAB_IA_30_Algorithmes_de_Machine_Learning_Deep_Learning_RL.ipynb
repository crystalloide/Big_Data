## 16. Q-Learning

```
grid_size = 5
goal = (4,4)
obstacles = [(1,1),(2,2),(3,1)]
print("Grille", grid_size, "x", grid_size, "| but:", goal, "| obstacles:", obstacles)
```

```
import random
Q = np.zeros((grid_size, grid_size, 4))
actions = [(-1,0),(1,0),(0,-1),(0,1)]
def step(state, a):
    dx, dy = actions[a]
    ns = (max(0,min(grid_size-1,state[0]+dx)), max(0,min(grid_size-1,state[1]+dy)))
    if ns in obstacles: return state, -5
    if ns == goal: return ns, 10
    return ns, -1
alpha, gamma, eps = 0.1, 0.9, 0.2
for ep in range(500):
    s = (0,0)
    for t in range(50):
        a = random.randint(0,3) if random.random()<eps else np.argmax(Q[s[0],s[1]])
        ns, r = step(s, a)
        Q[s[0],s[1],a] += alpha*(r + gamma*np.max(Q[ns[0],ns[1]]) - Q[s[0],s[1],a])
        s = ns
        if s == goal: break
print("Q-table entrainee. Valeur max etat (0,0):", Q[0,0].max())
```

## 17. SARSA

```
grid_size = 5
goal = (4,4)
obstacles = [(0,3),(2,1),(3,3)]
print("Grille SARSA prete | obstacles:", obstacles)
```

```
Qs = np.zeros((grid_size, grid_size, 4))
def step(state, a):
    dx, dy = actions[a]
    ns = (max(0,min(grid_size-1,state[0]+dx)), max(0,min(grid_size-1,state[1]+dy)))
    if ns in obstacles: return ns, -5
    if ns == goal: return ns, 10
    return ns, -1
alpha, gamma, eps = 0.1, 0.9, 0.2
for ep in range(500):
    s = (0,0)
    a = random.randint(0,3) if random.random()<eps else np.argmax(Qs[s[0],s[1]])
    for t in range(50):
        ns, r = step(s, a)
        na = random.randint(0,3) if random.random()<eps else np.argmax(Qs[ns[0],ns[1]])
        Qs[s[0],s[1],a] += alpha*(r + gamma*Qs[ns[0],ns[1],na] - Qs[s[0],s[1],a])
        s, a = ns, na
        if s == goal: break
print("SARSA entraine. Valeur max etat (0,0):", Qs[0,0].max())
```

## 22. Convolutional Neural Network (CNN)

```
def make_shape_dataset(n=500):
    X = np.zeros((n,28,28,1)); y = np.zeros(n, dtype=int)
    for i in range(n):
        img = np.zeros((28,28))
        if i % 2 == 0:
            cx, cy, r = 14,14,8
            yy, xx = np.ogrid[:28,:28]
            img[(xx-cx)**2+(yy-cy)**2 <= r**2] = 1
            y[i] = 0
        else:
            img[6:22,6:22] = 1
            y[i] = 1
        X[i,:,:,0] = img + np.random.normal(0,0.05,(28,28))
    return X, y
X, y = make_shape_dataset(500)
plt.imshow(X[0,:,:,0], cmap='gray'); plt.title("Exemple cercle"); plt.show()
```

```
from tensorflow.keras import layers
model = tf.keras.Sequential([
    layers.Conv2D(16, 3, activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(32, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
hist = model.fit(X, y, epochs=5, verbose=0)
print("Accuracy CNN:", hist.history['accuracy'][-1])
```

## 23. Recurrent Neural Network (RNN)

```
t = np.linspace(0, 50, 1000)
series = np.sin(t) + np.random.normal(0, 0.1, 1000)
def to_sequences(s, window=20):
    X, y = [], []
    for i in range(len(s)-window):
        X.append(s[i:i+window]); y.append(s[i+window])
    return np.array(X), np.array(y)
X, y = to_sequences(series)
X = X.reshape(-1,20,1)
print(X.shape, y.shape)
```

```
from tensorflow.keras import layers
model = tf.keras.Sequential([
    layers.SimpleRNN(32, input_shape=(20,1)),
    layers.Dense(1)
])
model.compile(optimizer='adam', loss='mse')
hist = model.fit(X, y, epochs=10, verbose=0)
print("Loss finale RNN:", hist.history['loss'][-1])
```

## 29. Markov Decision Process (MDP)

```
states = [0,1,2]; actions_mdp = [0,1]
P = {
 (0,0):[(0.7,0),(0.3,1)], (0,1):[(0.4,0),(0.6,1)],
 (1,0):[(0.5,1),(0.5,2)], (1,1):[(0.2,0),(0.8,2)],
 (2,0):[(1.0,2)], (2,1):[(0.6,1),(0.4,2)]
}
R = {0:5, 1:-1, 2:10}
print("MDP defini | etats:", states, "| recompenses:", R)
```

```
V = {s:0 for s in states}
gamma = 0.9
for _ in range(100):
    newV = {}
    for s in states:
        newV[s] = max(sum(p*(R[s2]+gamma*V[s2]) for p,s2 in P[(s,a)]) for a in actions_mdp)
    V = newV
print("Valeurs optimales des etats:", V)
```

## 30. Genetic Algorithms

```
import random
pop_size, gene_len = 50, 20
population = [np.random.randint(0,2,gene_len) for _ in range(pop_size)]
def fitness(ind): return ind.sum()
print("Population initiale, fitness moyenne:", np.mean([fitness(i) for i in population]))
```

```
for gen in range(100):
    scores = [fitness(i) for i in population]
    parents = [population[i] for i in np.argsort(scores)[-20:]]
    offspring = []
    for _ in range(pop_size):
        p1, p2 = random.sample(parents, 2)
        cut = random.randint(1, gene_len-1)
        child = np.concatenate([p1[:cut], p2[cut:]])
        if random.random() < 0.05:
            idx = random.randint(0, gene_len-1); child[idx] = 1-child[idx]
        offspring.append(child)
    population = offspring
print("Fitness finale moyenne apres evolution:", np.mean([fitness(i) for i in population]))
```
